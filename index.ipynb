## Challenge 1: Turtle Race

Ziel: Schreiben Sie ein Programm, bei dem mehrere Schildkröten in einem Wettrennen gegeneinander antreten. Die Funktion randint(Startwert, Endwert) aus dem Modul random liefert zufällige Werte zurück und hilft Ihnen, die Bewegung der Schildkröten zufälliger zu gestalten.

Anleitung:
- Schreiben Sie eine Funktion move_turtle(turtle), die eine gegebene Schildkröte um einen zufälligen Wert vorwärts bewegt
- Erzeugen Sie mehrere Schildkröten an der Startlinie
- Nutzen Sie eine Schleife, um die Schildkröten solange nach vorne zu bewegen, bis eine die Ziellinie erreicht
- Zeigen Sie den Gewinner an.


```python
import turtle
import random

def move_turtle(t):
    t.forward(random.randint(1, 10))


screen = turtle.Screen()
screen.title("Turtle Race")

# Erzeuge Turtles
turtle1 = turtle.Turtle()

# Positioniere Turtles
turtle1.penup()
turtle1.goto(-200, -40)
turtle1.pendown()

# Renne
race_on = True
while race_on:
    move_turtle(turtle1)
    if turtle1.xcor() >= 200:
        print(f"{turtle1.color()[0]} turtle wins!")
        race_on = False
        break


turtle.done()
```

Turtle Labyrinth

```python
import turtle

# Display aufbauen
screen = turtle.Screen()
screen.title("Labyrinth")
screen.setup(width=600, height=600)

# Eine turtle erzeugen, die nur das Labyrinth zeichnet
maze_drawer = turtle.Turtle()
maze_drawer.speed(0)  # so schnell wie möglich
maze_drawer.penup()

# Layout des Labyrinths
maze = [
    "1111111111",
    "1000000001",
    "1011111101",
    "1000000101",
    "1111111101",
    "1000000001",
    "1111111111"
]

# Größe der einzelnen Zellen festlegen
cell_size = 60 

# Labyrinth zeichnen
for y in range(len(maze)):
    for x in range(len(maze[y])):
        if maze[y][x] == '1':  # Mauern
            maze_drawer.goto(x * cell_size - 300, 300 - y * cell_size)
            maze_drawer.pendown()
            maze_drawer.fillcolor("black")
            maze_drawer.begin_fill()
            for _ in range(4):
                maze_drawer.forward(cell_size)
                maze_drawer.right(90)
            maze_drawer.end_fill()
            maze_drawer.penup()

# turtle zum zeichnen ausblenden
maze_drawer.hideturtle()

# Neue Schildkröte erzeugen, die als Spieler fungiert
robot = turtle.Turtle()
robot.shape("turtle")
robot.color("green")  # Farbe ändern
robot.penup()

# Startposition
robot.goto(-240 + cell_size // 2, 240 - cell_size // 2) 

# Position des Ziels
goal_x = cell_size - 260
goal_y = 260 - 5 * cell_size

# Ziel zeichnen
goal = turtle.Turtle()
goal.shape("circle")
goal.color("red")
goal.penup()
goal.goto(goal_x, goal_y)  # Position the goal at maze[5][2]

# Spieler immer um 1 Feld bewegen
def move_forward():
    # in welche Richtung schaut Spieler, wo ist vorne?
    heading = robot.heading()
    if heading == 0:
        new_x = robot.xcor() + cell_size
        new_y = robot.ycor()
    elif heading == 90:  # schaut nach oben
        new_x = robot.xcor()
        new_y = robot.ycor() + cell_size
    elif heading == 180:  # schaut nach links
        new_x = robot.xcor() - cell_size
        new_y = robot.ycor()
    elif heading == 270:  # schaut nach unten
        new_x = robot.xcor()
        new_y = robot.ycor() - cell_size

    # Kollisionserkennung
    if is_path(new_x, new_y):
        robot.goto(new_x, new_y)  # Falls möglich, gehe zu neuer Position
    
    # Ziel erreicht?
    if robot.distance(goal) < 30:  # Grenzwert für Zielerreichung
        print("Ziel erreicht!")

# Funktion zum links drehen
def turn_left():
    robot.left(90)

# Funktion zum rechts drehen
def turn_right():
    robot.right(90)

# Kollisionserkennung
def is_path(x, y):
    # Berechne Position im Grid
    grid_x = (x + 300) // cell_size
    grid_y = (300 - y) // cell_size
    # Grenzen und Pfad prüfen
    if 0 <= grid_x < len(maze[0]) and 0 <= grid_y < len(maze):
        return maze[grid_y][grid_x] == '0'
    return False

# Spielerbewegungen an Tastatur binden
screen.listen()
screen.onkey(move_forward, "Up")
screen.onkey(turn_left, "Left")
screen.onkey(turn_right, "Right")

# Fenster offen lassen
turtle.done()
```
